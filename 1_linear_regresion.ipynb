```python

import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Menyiapkan data dataset
bedrooms = np.array([1, 1, 2, 2, 3, 4, 4, 5, 5, 5])
house_price = np.array(
    [15000, 18000, 27000, 34000, 50000, 68000, 65000, 81000, 85000, 90000]
)

# 2. Mengubah bentuk data agar sesuai dengan kebutuhan scikit-learn
# Mengubah array 1D menjadi 2D (baris, 1 kolom)
bedrooms = bedrooms.reshape(-1, 1)

# 3. Membuat dan melatih model Linear Regression
linreg = LinearRegression()
linreg.fit(bedrooms, house_price)

# 4. Membuat visualisasi grafik scatter plot dan garis regresi
plt.scatter(bedrooms, house_price, color="blue", label="Data Aktual")
plt.plot(
    bedrooms, linreg.predict(bedrooms), color="red", label="Garis Regresi"
)
plt.xlabel("Jumlah Kamar Tidur (Bedrooms)")
plt.ylabel("Harga Rumah (House Price)")
plt.title("Analisis Regresi Linear Harga Rumah")
plt.legend()
plt.show()

# 5. Melakukan prediksi untuk rumah dengan 6 kamar tidur
prediksi_6 = linreg.predict([[6]])
print("Prediksi harga rumah untuk 6 kamar tidur:", prediksi_6[0])
```
